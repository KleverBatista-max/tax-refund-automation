# Obtencion de Tracking Id

```python
from bs4 import BeautifulSoup
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.support.ui import WebDriverWait

from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from bs4 import BeautifulSoup

options = Options()
options.add_argument("user-data-dir=C:\Users\YOU\AppData\\Local\\Google\\Chrome\\User Data")
options.add_argument("profile-directory=Profile 1")
driver = webdriver.Chrome(executable_path='C:\Users\YOU\Downloads\\chromedriver_win32(1)\\chromedriver.exe', options=options)

# Create a new Chrome driver instance
driver.get('https://www.amazon.com/gp/css/order-history')

# Get the HTML of the page
html = driver.page_source


soup = BeautifulSoup(html, 'html.parser')

divs = soup.find_all('div', {'class': 'yohtmlc-shipment-level-connections'})
primeros_a = []
for div in divs:
    button_general = div.find_all('a')
    for button in button_general:
        if button.text.strip() == "Track package":
            primeros_a.append(button)

ids = []
for a_tag in primeros_a:
    id = a_tag.get('id')
    ids.append(id)


tracking_ids = []
# Click en botones
for id in ids:
    element = WebDriverWait(driver, 10).until(
        EC.presence_of_element_located((By.ID, id))
    )
    element.click()
    
    WebDriverWait(driver, 10).until(EC.presence_of_element_located((By.CLASS_NAME, 'pt-delivery-card-trackingId')))
    
    # Actualizar la variable 'soup' con el nuevo HTML de la página
    html = driver.page_source
    soup = BeautifulSoup(html, 'html.parser')
    
    # Buscar los elementos 'trackings' en el nuevo HTML de la página
    trackings = soup.find_all('div', {'class': 'pt-delivery-card-trackingId'})
    for order in trackings:
        tracking_id = order.text.strip().split(': ')[1]
        tracking_ids.append(tracking_id)
    
    driver.back()

print(tracking_ids)
    



```
